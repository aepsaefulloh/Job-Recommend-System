# 📘 Supervised Learning

## 1. Definisi
Supervised Learning adalah pendekatan **machine learning** di mana model dilatih menggunakan data yang sudah memiliki **label**.  Setiap data pelatihan terdiri dari **input (fitur)** dan **output (target)** yang diketahui.

<br>

Tujuan algoritma adalah mempelajari hubungan antara input dan output sehingga mampu memprediksi output untuk data baru yang belum pernah dilihat.


<br>


## 2. Konsep Dasar
- **Input (X)** → fitur yang digunakan untuk prediksi, misalnya ukuran bunga (panjang sepal, lebar sepal, panjang petal, lebar petal).  
- **Output (y)** → label atau target yang ingin diprediksi, misalnya spesies bunga.  
- **Model** → algoritma yang digunakan untuk memetakan X → y.  
- **Training** → proses melatih model menggunakan data berlabel.  
- **Testing** → menguji model pada data baru untuk mengukur performa.


<br>


## 3. Proses Umum
1. **Kumpulkan Data** → data berlabel (X, y).  
2. **Bagi Data** → training set & testing set.  
3. **Pilih Algoritma** → misalnya Decision Tree, KNN, Logistic Regression, SVM, dsb.  
4. **Latih Model** → gunakan training set.  
5. **Evaluasi Model** → uji dengan testing set.  
6. **Prediksi** → gunakan model untuk memprediksi output data baru.



<br>

## 4. Contoh & Penerapan
Supervised learning digunakan dalam berbagai aplikasi nyata, di antaranya:  

- **Klasifikasi** → memprediksi kategori, misalnya email spam atau non-spam.  
- **Regresi** → memprediksi nilai numerik, misalnya harga rumah berdasarkan luas tanah dan lokasi.  
- **Penerapan umum**:  
  - Sistem rekomendasi film.  
  - Deteksi penyakit dari hasil medis.  
  - Prediksi harga saham.  
  - Pengenalan wajah atau suara.

## Definisi, Konsep Dasar, Proses Umum, Contoh & Penerapan Unsupervised Learning

<br>


### 1. **Definisi Unsupervised Learning**
<p>Unsupervised Learning (Pembelajaran Tanpa Pengawasan) adalah jenis algoritma pembelajaran mesin di mana model dilatih pada data yang tidak memiliki label. Dalam Unsupervised Learning, algoritma mencoba menemukan pola atau struktur tersembunyi dalam data yang tidak terlabel. Model ini bertujuan untuk membuat kesimpulan dari data tanpa panduan dari label atau hasil yang diketahui.<p>

<br>


### 2. **Konsep Dasar**
- **Data Tanpa Label**: Unsupervised learning menggunakan data yang tidak memiliki label atau kategori yang ditetapkan sebelumnya.
- **Tujuan**: Tujuan utama dari unsupervised learning adalah untuk menemukan struktur atau pola yang tersembunyi dalam data. Algoritma ini mencoba untuk mengelompokkan atau mereduksi dimensi data tanpa adanya penanda atau kategori yang ditetapkan.
- **Pengelompokan & Reduksi Dimensi**: Dua teknik utama yang digunakan dalam unsupervised learning adalah pengelompokan (clustering) dan reduksi dimensi (dimensionality reduction).

<br>


### 3. **Proses Umum Unsupervised Learning**
1. **Pengumpulan Data**: Mengumpulkan data tanpa label yang mencakup berbagai fitur atau variabel.
2. **Pra-pemrosesan Data**: Melakukan pembersihan dan persiapan data, seperti menghilangkan data yang hilang atau tidak konsisten.
3. **Pemilihan Algoritma**: Memilih algoritma yang tepat, seperti K-Means, DBSCAN, atau PCA (Principal Component Analysis).
4. **Modeling**: Melatih model menggunakan data yang telah diproses untuk menemukan pola atau struktur tersembunyi.
5. **Evaluasi dan Interpretasi**: Menginterpretasikan hasil yang diperoleh, seperti pengelompokan atau pengurangan dimensi, dan menganalisis pola atau informasi yang ditemukan.
6. **Aplikasi dan Penggunaan**: Menerapkan hasil dari unsupervised learning untuk membuat keputusan atau untuk analisis lebih lanjut.

<br>


### 4. **Contoh Unsupervised Learning**
- **K-Means Clustering**: Algoritma pengelompokan yang membagi data menjadi beberapa grup berdasarkan kedekatannya. Setiap grup atau kluster berisi data yang lebih mirip satu sama lain dibandingkan dengan data dari kluster lain.
- **Hierarchical Clustering**: Mengelompokkan data dalam bentuk pohon hierarki, yang menggambarkan hubungan antara data dengan cara berlapis.
- **PCA (Principal Component Analysis)**: Digunakan untuk reduksi dimensi, yaitu mengurangi jumlah variabel dalam data sambil tetap mempertahankan informasi yang paling penting.
- **Anomaly Detection**: Mengidentifikasi pola atau kejadian yang tidak biasa dalam data, yang dapat membantu mendeteksi penyimpangan atau anomali.

<br>


### 5. **Penerapan Unsupervised Learning**
Unsupervised learning banyak diterapkan dalam berbagai bidang, di antaranya:
- **Segmentasi Pasar**: Mengelompokkan pelanggan berdasarkan pola perilaku mereka, yang dapat digunakan untuk pemasaran yang lebih efektif.
- **Deteksi Penipuan**: Mengidentifikasi transaksi atau perilaku yang tidak biasa dalam sistem keuangan.
- **Rekomendasi Produk**: Menggunakan pengelompokan pengguna yang mirip untuk memberikan rekomendasi produk atau layanan.
- **Analisis Gambar**: Mengidentifikasi pola dalam gambar atau video untuk aplikasi seperti pengenalan objek atau pemetaan.
- **Reduksi Dimensi**: Meningkatkan efisiensi penyimpanan dan pemrosesan data besar dengan mengurangi jumlah fitur dalam dataset.


<br>

Unsupervised learning memungkinkan kita untuk menemukan pola yang tidak terlihat sebelumnya dalam data besar tanpa membutuhkan label, yang sangat berguna untuk eksplorasi data dan analisis yang lebih mendalam.

# Preprocessing: Scaling Data

Sebelum melakukan *clustering*, langkah penting yang harus dilakukan adalah **standarisasi data**.  
Hal ini dilakukan menggunakan **StandardScaler** dari Scikit-Learn.

## Mengapa Perlu Scaling?

- Fitur **TotalSpending** dan **PurchaseFrequency** mungkin memiliki skala yang berbeda.  
- Algoritma **K-Means** menggunakan jarak **Euclidean** untuk mengukur kedekatan antar data.  
- Jika skala fitur berbeda jauh, fitur dengan nilai besar akan lebih dominan dan menyebabkan hasil clustering tidak akurat.  

## Solusi

Dengan **StandardScaler**, setiap fitur akan diubah sehingga memiliki:  
- Rata-rata = 0  
- Standar deviasi = 1  

```python
# K-Means sangat sensitif terhadap skala fitur.
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_df)

print("\nData setelah Scaling (beberapa baris pertama):\n", X_scaled[:5])
```

# Menentukan Jumlah Cluster (Metode Elbow)

Untuk menentukan jumlah cluster yang optimal, digunakan **Metode Elbow**.  
Metode ini melibatkan perhitungan nilai **inertia** untuk berbagai jumlah cluster, kemudian divisualisasikan dalam bentuk grafik.

<br>

## Langkah-Langkah:

1. Menjalankan algoritma **K-Means** dengan jumlah cluster (`k`) dari 1 hingga 10.  
2. Menghitung **inertia**, yaitu total jarak kuadrat semua titik ke pusat cluster masing-masing.  
3. Membuat grafik hubungan antara jumlah cluster (k) dengan nilai inertia.  
4. Titik siku (*elbow point*) pada grafik dipilih sebagai jumlah cluster optimal.

```python
inertia_values = []
k_range = range(1, 11)

print("\nMencari jumlah cluster optimal (Metode Elbow)...")
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    inertia_values.append(kmeans.inertia_)

plt.figure(figsize=(8, 6))
plt.plot(k_range, inertia_values, marker='o')
plt.title('Metode Elbow untuk Jumlah Cluster Optimal')
plt.xlabel('Jumlah Cluster (k)')
plt.ylabel('Inersia')
plt.xticks(k_range)
plt.grid(True)
plt.show()

# Pilih k optimal (misalnya k=3 sesuai data sintetis)
optimal_k = 3
print(f"\nMemilih jumlah cluster optimal: k = {optimal_k}")
```

# Menerapkan K-Means dengan Cluster Optimal

Setelah jumlah cluster optimal ditentukan dengan **Metode Elbow** (misalnya `k = 3`),  
langkah selanjutnya adalah menjalankan algoritma **K-Means** untuk melakukan pengelompokan data.

<br>

## Penjelasan Proses

- Menjalankan **K-Means** dengan jumlah cluster = 3.  
- Setiap pelanggan akan diberikan **label cluster** (misalnya 0, 1, atau 2).  
- Algoritma juga menghasilkan **centroid**, yaitu titik pusat dari masing-masing cluster.  
- Hasil ini akan digunakan untuk analisis pola belanja tiap segmen pelanggan.  

```python
kmeans = KMeans(n_clusters=optimal_k, random_state=42, n_init=10)
kmeans.fit(X_scaled)

# Label segmen untuk setiap pelanggan
labels_segmen = kmeans.labels_
centroids_scaled = kmeans.cluster_centers_

print(f"\nK-Means selesai. Setiap pelanggan diberi label dari 0 hingga {optimal_k-1}.")
print("Contoh beberapa label segmen pertama:", labels_segmen[:20])
```
